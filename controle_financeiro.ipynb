```python
import pandas as pd
import numpy as np
import openpyxl
import plotly.express as px

print("Bibliotecas importadas com sucesso!")
```

```python

from openpyxl import Workbook

# Criar uma nova planilha
wb = Workbook()

# Definir as abas
abas = [
    "Dashboard",
    "Lançamentos",
    "Orçamento",
    "Metas",
    "Configurações"
]

# Remover a aba padrão
wb.remove(wb.active)

# Criar as abas do projeto
for aba in abas:
    wb.create_sheet(aba)

print(wb.sheetnames)
```

```python

# Identidade visual do projeto

cores = {
    "fundo": "0F172A",
    "painel": "1E293B",
    "receitas": "10B981",
    "despesas": "EF4444",
    "destaque": "38BDF8",
    "texto": "F8FAFC"
}

print(cores)
```

```python

from openpyxl.styles import Font, PatternFill, Alignment

# Selecionar a aba Dashboard
ws = wb["Dashboard"]

# Ocultar linhas de grade
ws.sheet_view.showGridLines = False

# Aplicar fundo escuro
for linha in ws.iter_rows(
    min_row=1, max_row=35,
    min_col=1, max_col=16
):
    for celula in linha:
        celula.fill = PatternFill(
            "solid", fgColor=cores["fundo"]
        )

# Criar título
ws.merge_cells("B2:O3")

ws["B2"] = "CONTROLE FINANCEIRO PRO 2.0"

ws["B2"].font = Font(
    name="Calibri",
    size=24,
    bold=True,
    color=cores["texto"]
)

ws["B2"].alignment = Alignment(
    horizontal="center",
    vertical="center"
)

print("Dashboard inicial criado!")
```
