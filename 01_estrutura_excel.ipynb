```python
import pandas as pd
import numpy as np
import openpyxl
import plotly.express as px

print("Bibliotecas importadas com sucesso!")
```

```python
# 1. Importação das bibliotecas

from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Alignment
from pathlib import Path


# 2. Criação do arquivo Excel

wb = Workbook()

abas = [
    "Dashboard",
    "Lançamentos",
    "Orçamento",
    "Metas",
    "Configurações"
]

for aba in abas:
    wb.create_sheet(aba)

wb.remove(wb["Sheet"])


# 3. Identidade visual

cores = {
    "fundo": "0F172A",
    "painel": "1E293B",
    "receitas": "10B981",
    "despesas": "EF4444",
    "destaque": "38BDF8",
    "texto": "F8FAFC"
}


# 4. Configuração do Dashboard

ws = wb["Dashboard"]

ws.sheet_view.showGridLines = False

for linha in ws["A1:P35"]:
    for celula in linha:
        celula.fill = PatternFill(
            fill_type="solid",
            fgColor=cores["fundo"]
        )


# 5. Título principal

ws.merge_cells("B2:O3")

ws["B2"] = "CONTROLE FINANCEIRO PRO 2.0"

ws["B2"].font = Font(
    name="Calibri",
    size=24,
    bold=True,
    color=cores["texto"]
)

ws["B2"].alignment = Alignment(
    horizontal="center",
    vertical="center"
)


# 6. Salvamento do arquivo

Path("output").mkdir(exist_ok=True)

wb.save("output/Controle_Financeiro_Pro.xlsx")

print("Arquivo Excel criado com sucesso!")
```
