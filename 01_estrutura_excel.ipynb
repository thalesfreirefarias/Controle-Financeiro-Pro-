```python
# ==========================================
# CONTROLE FINANCEIRO PRO 2.0
# DIA 1 - ESTRUTURA E IDENTIDADE VISUAL
# ==========================================


# 1. Importação das bibliotecas

import requests
import base64

from io import BytesIO
from google.colab import userdata
from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Alignment


# 2. Criação do arquivo Excel

wb = Workbook()

abas = [
    "Dashboard",
    "Lançamentos",
    "Orçamento",
    "Metas",
    "Configurações"
]

for aba in abas:
    wb.create_sheet(aba)

wb.remove(wb["Sheet"])


# 3. Identidade visual

cores = {
    "fundo": "0F172A",
    "painel": "1E293B",
    "receitas": "10B981",
    "despesas": "EF4444",
    "destaque": "38BDF8",
    "texto": "F8FAFC"
}


# 4. Configuração do Dashboard

ws = wb["Dashboard"]

ws.sheet_view.showGridLines = False

for linha in ws["A1:P35"]:
    for celula in linha:
        celula.fill = PatternFill(
            fill_type="solid",
            fgColor=cores["fundo"]
        )


# 5. Título principal

ws.merge_cells("B2:O3")

ws["B2"] = "CONTROLE FINANCEIRO PRO 2.0"

ws["B2"].font = Font(
    name="Calibri",
    size=24,
    bold=True,
    color=cores["texto"]
)

ws["B2"].alignment = Alignment(
    horizontal="center",
    vertical="center"
)


# 6. Configuração da conexão com o GitHub

token = userdata.get("GITHUB_TOKEN")

repositorio = "thalesfreirefarias/Controle-Financeiro-Pro-"

caminho = "output/Controle_Financeiro_Pro.xlsx"

url = (
    f"https://api.github.com/repos/"
    f"{repositorio}/contents/{caminho}"
)

headers = {
    "Authorization": f"Bearer {token}",
    "Accept": "application/vnd.github+json"
}


# 7. Salvar o Excel em memória

arquivo = BytesIO()

wb.save(arquivo)


# 8. Conversão do arquivo para Base64

conteudo = base64.b64encode(
    arquivo.getvalue()
).decode("utf-8")


# 9. Verificação do arquivo no GitHub

consulta = requests.get(
    url,
    headers=headers,
    params={"ref": "main"},
    timeout=30
)

payload = {
    "message": "feat: publish Excel workbook",
    "content": conteudo,
    "branch": "main"
}

if consulta.status_code == 200:

    payload["sha"] = consulta.json()["sha"]

elif consulta.status_code != 404:

    consulta.raise_for_status()


# 10. Publicação automática no GitHub

resposta = requests.put(
    url,
    headers=headers,
    json=payload,
    timeout=30
)

resposta.raise_for_status()

print("Excel publicado no GitHub com sucesso!")

print("\nAbas criadas:")
print(wb.sheetnames)

print("\nEndereço do arquivo:")
print(resposta.json()["content"]["html_url"])
```
